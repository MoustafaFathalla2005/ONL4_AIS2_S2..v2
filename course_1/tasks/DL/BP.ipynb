# Backpropagation Task 

notebook follows the exact steps required :

1. Compute net input for h1  
2. Compute output of h1 using sigmoid  
3. Repeat for h2  
4. Compute outputs for output layer  
5. Calculate total error  
6. Perform backward pass (update weights using d notation)  
7. Train for 10,000 iterations  
8. Final prediction

## Step 1 & 2 — Hidden Neuron h1

net_h1 = w1*x1 + w2*x2 + b1  
net_h1 = 0.15*0.05 + 0.20*0.10 + 0.35 = 0.3775  

out_h1 = sigmoid(net_h1)  
out_h1 = 0.593269992

## Step 3 — Hidden Neuron h2

net_h2 = w3*x1 + w4*x2 + b1  
net_h2 = 0.3925  

out_h2 = 0.596884378

## Step 4 — Output Layer

net_o1 = 1.1059 → out_o1 = 0.75136507  
net_o2 = 1.2249 → out_o2 = 0.772928465

## Step 5 — Total Error

E = 1/2 (t1 - o1)^2 + 1/2 (t2 - o2)^2  

E = 0.298371

## Step 6 — Backward Pass

Using chain rule (d notation):

### Output Layer Example

dE/dw5 = (dE/dout_o1)(dout_o1/dnet_o1)(dnet_o1/dw5)

Where:
- dE/dout_o1 = out_o1 - target_o1  
- dout_o1/dnet_o1 = out_o1(1 - out_o1)  
- dnet_o1/dw5 = out_h1  

---

### Hidden Layer

dE/dw1 = (dE/dout_h1)(dout_h1/dnet_h1)(dnet_h1/dw1)

Weights are updated using:

w_new = w_old - lr * dE/dw

```python
import numpy as np
```

```python
class NeuralNetwork:
    """
    Two-layer neural network (2-2-2) using sigmoid activation
    Implements forward pass, backward pass, and training.
    """

    def __init__(self, lr=0.5):
        self.lr = lr

        # Initial weights
        self.w1, self.w2, self.w3, self.w4 = 0.15, 0.20, 0.25, 0.30
        self.w5, self.w6, self.w7, self.w8 = 0.40, 0.45, 0.50, 0.55

        # Biases
        self.b1, self.b2 = 0.35, 0.60

    def sigmoid(self, x):
        """Sigmoid activation function
        x: input value
        returns: sigmoid output
        """
        return 1 / (1 + np.exp(-x))

    def d_sigmoid(self, x):
        """Derivative of sigmoid
        x: sigmoid output
        returns: derivative value"""
        return x * (1 - x)

    def forward(self, x1, x2):
        """Perform forward propagation
        x1, x2: input values
        returns: output values (o1, o2)"""

        self.net_h1 = self.w1*x1 + self.w2*x2 + self.b1
        self.out_h1 = self.sigmoid(self.net_h1)

        self.net_h2 = self.w3*x1 + self.w4*x2 + self.b1
        self.out_h2 = self.sigmoid(self.net_h2)

        self.net_o1 = self.w5*self.out_h1 + self.w6*self.out_h2 + self.b2
        self.out_o1 = self.sigmoid(self.net_o1)

        self.net_o2 = self.w7*self.out_h1 + self.w8*self.out_h2 + self.b2
        self.out_o2 = self.sigmoid(self.net_o2)

        return self.out_o1, self.out_o2

    def compute_error(self, t1, t2):
        """Compute total error
        t1, t2: target values
        returns: total error"""
        return 0.5*((t1 - self.out_o1)**2 + (t2 - self.out_o2)**2)

    def backward(self, x1, x2, t1, t2):
        """Perform backpropagation using d notation
        x1, x2: input values
        t1, t2: target values"""

        # Output layer
        d_o1 = (self.out_o1 - t1) * self.d_sigmoid(self.out_o1)
        d_o2 = (self.out_o2 - t2) * self.d_sigmoid(self.out_o2)

        # Update output weights
        self.w5 -= self.lr * d_o1 * self.out_h1
        self.w6 -= self.lr * d_o1 * self.out_h2
        self.w7 -= self.lr * d_o2 * self.out_h1
        self.w8 -= self.lr * d_o2 * self.out_h2
        self.b2 -= self.lr * (d_o1 + d_o2)

        # Hidden layer
        d_h1 = (d_o1*self.w5 + d_o2*self.w7) * self.d_sigmoid(self.out_h1)
        d_h2 = (d_o1*self.w6 + d_o2*self.w8) * self.d_sigmoid(self.out_h2)

        # Update hidden weights
        self.w1 -= self.lr * d_h1 * x1
        self.w2 -= self.lr * d_h1 * x2
        self.w3 -= self.lr * d_h2 * x1
        self.w4 -= self.lr * d_h2 * x2
        self.b1 -= self.lr * (d_h1 + d_h2)
```

```python
# Data
x1, x2 = 0.05, 0.10
t1, t2 = 0.01, 0.99

# Model
nn = NeuralNetwork(lr=0.5)

# Training
for epoch in range(10000):
    nn.forward(x1, x2)
    nn.backward(x1, x2, t1, t2)
```

```python
pred1, pred2 = nn.forward(x1, x2)

print("Final Prediction:")
print(round(pred1, 6), round(pred2, 6))
print("Target:", t1, t2)
print("Final Weights:")
print(f"w1: {nn.w1:.6f}, w2: {nn.w2:.6f}, w3: {nn.w3:.6f}, w4: {nn.w4:.6f}")
print(f"w5: {nn.w5:.6f}, w6: {nn.w6:.6f}, w7: {nn.w7:.6f}, w8: {nn.w8:.6f}")
print(f"b1: {nn.b1:.6f}, b2: {nn.b2:.6f}")
```
