https://leetcode.com/problems/two-sum/description/



Example :

Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
Example 2:

Input: nums = [3,2,4], target = 6
Output: [1,2]
Example 3:

Input: nums = [3,3], target = 6
Output: [0,1]

```python
# more complexity  o(n^2)
def twoSum1(nums, target) :
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                return [i, j]
```

```python
def twoSum2(nums, target) :
    num_map = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in num_map:
            return [num_map[complement], i]
        num_map[num] = i
```

```python
# use case 
def run_test_case(nums, target, case_name=""):
    print(f"\n{case_name}")
    print(f"Input: nums={nums}, target={target}")
    
    r1 = twoSum1(nums, target)
    r2 = twoSum2(nums, target)
    
    print(f"(Function 1): {r1}")
    print(f"(Function 2): {r2}")

print("=" * 50)
print("TWO SUM TEST CASES")
print("=" * 50)

run_test_case([2,7,11,15], 9, "Example 1")
run_test_case([3,2,4], 6, "Example 2")
run_test_case([3,3], 6, "Example 3")
run_test_case([-1,-2,-3,-4], -5, "Negative Numbers")
run_test_case([1,2,3,4], 10, "No Solution")
```

https://leetcode.com/problems/add-two-numbers/description/

Example :


Input: l1 = [2,4,3], l2 = [5,6,4]
Output: [7,0,8]
Explanation: 342 + 465 = 807.
Example 2:

Input: l1 = [0], l2 = [0]
Output: [0]
Example 3:

Input: l1 = [9,9,9,9,9,9,9], l2 = [9,9,9,9]
Output: [8,9,9,9,0,0,0,1]

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def list_to_linkedlist(arr):
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head

def print_linkedlist(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    print(result)
```

```python

def addTwoNumbers_1(l1, l2):
    num1 = 0
    num2 = 0
    multiplier = 1
    
    while l1:
        num1 += l1.val * multiplier
        multiplier *= 10
        l1 = l1.next
    
    multiplier = 1
    while l2:
        num2 += l2.val * multiplier
        multiplier *= 10
        l2 = l2.next
    
    total = num1 + num2  
    
    dummy = ListNode(0)
    curr = dummy
    
    if total == 0:
        return ListNode(0)
    
    while total > 0:
        curr.next = ListNode(total % 10)
        curr = curr.next
        total //= 10
    
    return dummy.next

def addTwoNumbers_2(l1, l2):
    dummy = ListNode(0)
    curr = dummy
    carry = 0
    
    while l1 or l2 or carry:
        total = (l1.val if l1 else 0) + (l2.val if l2 else 0) + carry
        curr.next = ListNode(total % 10)
        carry = total // 10
        curr = curr.next
        l1 = l1.next if l1 else None
        l2 = l2.next if l2 else None
    
    return dummy.next
```

```python
# use case
def run_test_case(arr1, arr2, case_name=""):
    print(f"{case_name}")
    print(f"Input: {arr1} + {arr2}")
    l1 = list_to_linkedlist(arr1)
    l2 = list_to_linkedlist(arr2)
    r1 = addTwoNumbers_1(l1, l2)
    r2 = addTwoNumbers_2(l1, l2)
    
    print("Function 1: ", end="")
    print_linkedlist(r1)
    print("Function 2: ", end="")
    print_linkedlist(r2)

run_test_case([2,4,3], [5,6,4], "Example 1")
run_test_case([9,9,9,9,9,9,9], [9,9,9,9], "Example 2")
run_test_case([0], [0], "Zeros")
```

https://leetcode.com/problems/roman-to-integer/description/


Example :

Input: s = "III"
Output: 3
Explanation: III = 3.
Example 2:

Input: s = "LVIII"
Output: 58
Explanation: L = 50, V= 5, III = 3.
Example 3:

Input: s = "MCMXCIV"
Output: 1994
Explanation: M = 1000, CM = 900, XC = 90 and IV = 4.

```python
def romanToInt1(s):
        roman_values = {
            'I': 1,
            'V': 5,
            'X': 10,
            'L': 50,
            'C': 100,
            'D': 500,
            'M': 1000
        }
        
        total = 0
        n = len(s)
        
        for i in range(n):
            if i < n - 1 and roman_values[s[i]] < roman_values[s[i + 1]]:
                total -= roman_values[s[i]]
            else:
                total += roman_values[s[i]]
        
        return total
```

```python
def romanToInt2(s):
    values = {'I':1, 'V':5, 'X':10, 'L':50, 'C':100, 'D':500, 'M':1000}
    total = 0
    
    for i in range(len(s)):
        if i+1 < len(s) and values[s[i]] < values[s[i+1]]:
            total -= values[s[i]]
        else:
            total += values[s[i]]
    
    return total
```

```python
def test_roman(roman, expected):
    r1 = romanToInt1(roman)
    r2 = romanToInt2(roman)
    print(f"{roman} → r1: {r1}, r2: {r2} (expected: {expected})")
    

print(" SIMPLE TEST CASES")
print("-" * 50)
test_roman("III", 3)
test_roman("IV", 4)
test_roman("IX", 9)
test_roman("LVIII", 58)
test_roman("MCMXCIV", 1994)
test_roman("MMMCMXCIX", 3999)
```

https://leetcode.com/problems/search-insert-position/description/


Example 1:

Input: nums = [1,3,5,6], target = 5
Output: 2
Example 2:

Input: nums = [1,3,5,6], target = 2
Output: 1
Example 3:

Input: nums = [1,3,5,6], target = 7
Output: 4
 

```python
def searchInsert1(nums, target):
    l, r = 0, len(nums) - 1
    while l <= r:
        m = (l + r) // 2
        if nums[m] == target:
            return m
        if nums[m] < target:
            l = m + 1
        else:
            r = m - 1
    return l
```

```python
def test(nums, target, expected):
    result1 = searchInsert1(nums, target)
    # result2 = searchInsert2(nums, target)
    # print(f"nums={nums}, target={target} → Result 1 : {result1}, Result 2 : {result2} (expected: {expected})")
    print(f"nums={nums}, target={target} → Result 1 : {result1} (expected: {expected})")

print("SIMPLE TEST CASES")
print("-" * 40)
test([1,3,5,6], 5, 2)
test([1,3,5,6], 2, 1)
test([1,3,5,6], 7, 4)
test([1,3,5,6], 0, 0)
test([1], 0, 0)
test([1], 2, 1)
test([1,3,5], 4, 2)
```
