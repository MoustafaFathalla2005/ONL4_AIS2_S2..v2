Task1

https://leetcode.com/problems/basic-calculator-ii/description/?envType=problem-list-v2&envId=math


Example 1:

Input: s = "3+2*2"
Output: 7
Example 2:

Input: s = " 3/2 "
Output: 1
Example 3:

Input: s = " 3+5 / 2 "
Output: 5

```python
def calculate1(s: str) -> int:
    """
    Solution 1: Using Stack
    Time: O(n) | Space: O(n)
    
    How it works:
        1. Iterate through each character
        2. Build numbers from digits
        3. When encountering operator:
           - For +: push number to stack
           - For -: push negative number
           - For * or /: pop last number, compute, push result
        4. Sum all numbers in stack
    """
    stack = []
    num = 0
    sign = '+'
    i = 0
    
    while i < len(s):
        if s[i].isdigit():
            num = num * 10 + int(s[i])
        
        if (not s[i].isdigit() and s[i] != ' ') or i == len(s) - 1:
            if sign == '+':
                stack.append(num)
            elif sign == '-':
                stack.append(-num)
            elif sign == '*':
                stack.append(stack.pop() * num)
            elif sign == '/':
                stack.append(int(stack.pop() / num))
            sign = s[i]
            num = 0
        i += 1
    
    return sum(stack)


def calculate2(s: str) -> int:
    """
    Solution 2: Without Stack (Space Optimized)
    Time: O(n) | Space: O(1)
    
    How it works:
        1. Track current number, last number, and running result
        2. For +: add last number to result, set last = num
        3. For -: add last to result, set last = -num
        4. For *: multiply last by num
        5. For /: divide last by num
        6. Finally add last to result
    """
    num = 0
    last = 0
    result = 0
    sign = '+'
    i = 0
    
    while i < len(s):
        if s[i].isdigit():
            num = num * 10 + int(s[i])
        
        if (not s[i].isdigit() and s[i] != ' ') or i == len(s) - 1:
            if sign == '+':
                result += last
                last = num
            elif sign == '-':
                result += last
                last = -num
            elif sign == '*':
                last = last * num
            elif sign == '/':
                last = int(last / num)
            sign = s[i]
            num = 0
        i += 1
    
    result += last
    return result


def test_calculator():
    """Test both calculator solutions"""
    test_cases = [
        ("3+2*2", 7),
        (" 3/2 ", 1),
        (" 3+5 / 2 ", 5),
        ("42", 42),
        ("1+1+1", 3),
        ("10*2+3", 23),
        ("100-50/2", 75),
    ]
    
    print("=" * 60)
    print("🔷 BASIC CALCULATOR II - COMPARISON")
    print("=" * 60)
    
    for expr, expected in test_cases:
        r1 = calculate1(expr)
        r2 = calculate2(expr)
        print(f"\n📝 Expression: '{expr}'")
        print(f"   Expected: {expected}")
        print(f"   Solution 1 (Stack):     {r1} {'✅' if r1 == expected else '❌'}")
        print(f"   Solution 2 (No Stack):  {r2} {'✅' if r2 == expected else '❌'}")
    
    print("\n" + "=" * 60)
```

Task2

https://leetcode.com/problems/coin-change/


Example 1:

Input: coins = [1,2,5], amount = 11
Output: 3
Explanation: 11 = 5 + 5 + 1
Example 2:

Input: coins = [2], amount = 3
Output: -1
Example 3:

Input: coins = [1], amount = 0
Output: 0

```python
def coinChange1(coins, amount):
    """
    Solution 1: Dynamic Programming (Bottom-Up)
    Time: O(amount * len(coins)) | Space: O(amount)
    
    How it works:
        1. Create dp array where dp[i] = min coins for amount i
        2. Initialize dp[0] = 0, others = infinity
        3. For each amount from 1 to target:
           Try each coin, if coin <= amount:
               dp[amount] = min(dp[amount], dp[amount - coin] + 1)
        4. If dp[amount] is still infinity, return -1
    
    Example: coins=[1,2,5], amount=11
    dp[0]=0
    dp[1]=1 (1)
    dp[2]=1 (2)
    dp[3]=2 (2+1)
    dp[4]=2 (2+2)
    dp[5]=1 (5)
    dp[6]=2 (5+1)
    dp[7]=2 (5+2)
    dp[8]=3 (5+2+1)
    dp[9]=3 (5+2+2)
    dp[10]=2 (5+5)
    dp[11]=3 (5+5+1)
    """
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)
    
    return dp[amount] if dp[amount] != float('inf') else -1


def coinChange2(coins, amount):
    """
    Solution 2: BFS (Breadth First Search)
    Time: O(amount * len(coins)) | Space: O(amount)
    
    How it works:
        1. Treat as shortest path problem
        2. Each node is an amount, each coin is an edge
        3. BFS guarantees minimum steps (coins count)
        4. Start from 0, explore all possible amounts
    """
    if amount == 0:
        return 0
    
    from collections import deque
    coins.sort()
    visited = [False] * (amount + 1)
    queue = deque([(0, 0)])  # (current_amount, steps)
    visited[0] = True
    
    while queue:
        current, steps = queue.popleft()
        
        for coin in coins:
            next_amount = current + coin
            
            if next_amount == amount:
                return steps + 1
            
            if next_amount < amount and not visited[next_amount]:
                visited[next_amount] = True
                queue.append((next_amount, steps + 1))
    
    return -1


def coinChange3(coins, amount):
    """
    Solution 3: DFS with Memoization (Top-Down DP)
    Time: O(amount * len(coins)) | Space: O(amount)
    
    How it works:
        1. Try all possibilities recursively
        2. Memoize results to avoid recomputation
        3. For each remaining amount, try all coins
    """
    memo = {}
    
    def dfs(remaining):
        if remaining < 0:
            return -1
        if remaining == 0:
            return 0
        if remaining in memo:
            return memo[remaining]
        
        min_coins = float('inf')
        for coin in coins:
            res = dfs(remaining - coin)
            if res != -1:
                min_coins = min(min_coins, res + 1)
        
        memo[remaining] = min_coins if min_coins != float('inf') else -1
        return memo[remaining]
    
    return dfs(amount)


def test_coin_change():
    """Test all coin change solutions"""
    test_cases = [
        ([1, 2, 5], 11, 3, "Example 1"),
        ([2], 3, -1, "Example 2"),
        ([1], 0, 0, "Example 3"),
        ([1, 3, 4], 6, 2, "4+1+1 or 3+3"),
        ([2, 5, 10, 1], 27, 4, "Multiple coins"),
        ([3, 5], 2, -1, "Impossible"),
    ]
    
    print("\n" + "=" * 70)
    print("🔷 COIN CHANGE - ALL SOLUTIONS COMPARISON")
    print("=" * 70)
    
    for coins, amount, expected, desc in test_cases:
        print(f"\n📝 Test: {desc}")
        print(f"   Coins: {coins}")
        print(f"   Amount: {amount}")
        print(f"   Expected: {expected}")
        print("-" * 60)
        
        r1 = coinChange1(coins, amount)
        r2 = coinChange2(coins, amount)
        r3 = coinChange3(coins, amount)
        
        print(f"   Solution 1 (DP):        {r1} {'✅' if r1 == expected else '❌'}")
        print(f"   Solution 2 (BFS):       {r2} {'✅' if r2 == expected else '❌'}")
        print(f"   Solution 3 (DFS Memo):  {r3} {'✅' if r3 == expected else '❌'}")
    
    print("\n" + "=" * 70)
```

Task3

https://leetcode.com/problems/validate-binary-search-tree/

Example 1:


Input: root = [2,1,3]
Output: true
Example 2:


Input: root = [5,1,4,null,null,3,6]
Output: false
Explanation: The root node's value is 5 but its right child's value is 4.
 

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def create_tree(values):
    """Create binary tree from list (level order)"""
    if not values:
        return None
    root = TreeNode(values[0])
    queue = [root]
    i = 1
    while queue and i < len(values):
        node = queue.pop(0)
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i])
            queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i])
            queue.append(node.right)
        i += 1
    return root


def isValidBST1(root) -> bool:
    """
    Solution 1: Inorder Traversal
    Time: O(n) | Space: O(h)
    
    How it works:
        1. Perform inorder traversal (left → root → right)
        2. In a valid BST, inorder traversal yields sorted values
        3. Track previous value, ensure current > previous
    """
    prev = None
    
    def inorder(node):
        nonlocal prev
        if not node:
            return True
        
        if not inorder(node.left):
            return False
        
        if prev is not None and node.val <= prev:
            return False
        prev = node.val
        
        return inorder(node.right)
    
    return inorder(root)


def isValidBST2(root) -> bool:
    """
    Solution 2: Range Check (Min/Max Validation)
    Time: O(n) | Space: O(h)
    
    How it works:
        1. Pass valid range (min, max) for each node
        2. Left child must be: min < left.val < current.val
        3. Right child must be: current.val < right.val < max
        4. Recursively validate all nodes
    """
    def validate(node, min_val, max_val):
        if not node:
            return True
        
        if (min_val is not None and node.val <= min_val) or \
           (max_val is not None and node.val >= max_val):
            return False
        
        return validate(node.left, min_val, node.val) and \
               validate(node.right, node.val, max_val)
    
    return validate(root, None, None)


def test_bst():
    """Test both BST validation solutions"""
    test_cases = [
        ([2, 1, 3], True, "Valid BST"),
        ([5, 1, 4, None, None, 3, 6], False, "Invalid BST (4 in right subtree)"),
        ([1, 1], False, "Duplicate values"),
        ([], True, "Empty tree"),
        ([10, 5, 15, None, None, 6, 20], False, "Invalid (6 < 10 but in right)"),
    ]
    
    print("\n" + "=" * 60)
    print("🔷 VALIDATE BST - COMPARISON")
    print("=" * 60)
    
    for values, expected, desc in test_cases:
        root = create_tree(values)
        r1 = isValidBST1(root)
        r2 = isValidBST2(root)
        print(f"\n📝 Test: {desc}")
        print(f"   Tree: {values}")
        print(f"   Expected: {expected}")
        print(f"   Solution 1 (Inorder):     {r1} {'✅' if r1 == expected else '❌'}")
        print(f"   Solution 2 (Range Check): {r2} {'✅' if r2 == expected else '❌'}")
    
    print("\n" + "=" * 60)
```

Task4

https://leetcode.com/problems/maximum-depth-of-binary-tree/

Example 1:


Input: root = [3,9,20,null,null,15,7]
Output: 3
Example 2:

Input: root = [1,null,2]
Output: 2

```python
def maxDepth1(root) -> int:
    """
    Solution 1: Recursion (DFS)
    Time: O(n) | Space: O(h)
    
    How it works:
        depth = 1 + max(depth(left), depth(right))
    """
    if not root:
        return 0
    return 1 + max(maxDepth1(root.left), maxDepth1(root.right))


def maxDepth2(root) -> int:
    """
    Solution 2: BFS (Level Order) using Queue
    Time: O(n) | Space: O(w)
    
    How it works:
        1. Process tree level by level
        2. Each level increments depth by 1
        3. Use queue to track nodes at current level
    """
    if not root:
        return 0
    
    from collections import deque
    queue = deque([root])
    depth = 0
    
    while queue:
        depth += 1
        for _ in range(len(queue)):
            node = queue.popleft()
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
    
    return depth


def maxDepth3(root) -> int:
    """
    Solution 3: Iterative DFS using Stack
    Time: O(n) | Space: O(h)
    
    How it works:
        1. Use stack to store (node, depth)
        2. Push root with depth 1
        3. Update max_depth when popping
        4. Push children with depth + 1
    """
    if not root:
        return 0
    
    stack = [(root, 1)]
    max_depth = 0
    
    while stack:
        node, depth = stack.pop()
        max_depth = max(max_depth, depth)
        
        if node.right:
            stack.append((node.right, depth + 1))
        if node.left:
            stack.append((node.left, depth + 1))
    
    return max_depth


def test_depth():
    """Test all depth solutions"""
    test_cases = [
        ([3, 9, 20, None, None, 15, 7], 3, "Balanced tree"),
        ([1, None, 2], 2, "Right skewed"),
        ([], 0, "Empty tree"),
        ([1], 1, "Single node"),
        ([1, 2, 3, 4, 5], 3, "Left skewed"),
    ]
    
    print("\n" + "=" * 60)
    print("🔷 MAXIMUM DEPTH - COMPARISON")
    print("=" * 60)
    
    for values, expected, desc in test_cases:
        root = create_tree(values)
        r1 = maxDepth1(root)
        r2 = maxDepth2(root)
        r3 = maxDepth3(root)
        print(f"\n📝 Test: {desc}")
        print(f"   Tree: {values}")
        print(f"   Expected: {expected}")
        print(f"   Recursion:   {r1} {'✅' if r1 == expected else '❌'}")
        print(f"   BFS:         {r2} {'✅' if r2 == expected else '❌'}")
        print(f"   DFS Stack:   {r3} {'✅' if r3 == expected else '❌'}")
    
    print("\n" + "=" * 60)
```
