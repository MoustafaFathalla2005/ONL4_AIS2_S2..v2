## 🧩 Problem 1 — Product of Array Except Self  

🔗 https://leetcode.com/problems/product-of-array-except-self/

### 💡 Idea
Compute product of all elements except current index  
without using division.

---

### ⚙️ Approaches

This problem requires computing the product of all elements in the array except the current index without using division.

There are two main approaches to solve this problem efficiently.

---

### Approach 1: Prefix and Suffix Arrays

In this method, the solution is divided into two stages.

First, a prefix product is computed for each index. This represents the product of all elements to the left of the current position.

Second, a suffix product is computed for each index. This represents the product of all elements to the right of the current position.

The final result for each index is obtained by multiplying the prefix and suffix values.

This approach is straightforward and easy to understand. However, it requires additional memory to store intermediate arrays.

---

### Approach 2: Optimized Space Solution

This approach improves memory usage by avoiding the use of extra arrays.

Instead of storing prefix and suffix arrays separately, the result array is used to store prefix values during the first pass.

A single variable is then used to maintain the suffix product during a reverse traversal of the array.

This approach maintains the same time complexity but reduces space usage significantly.

It is more efficient and preferred in interview settings.

---

| Approach | Idea | Space |
|---------|------|------|
| Prefix + Suffix | Build left & right arrays | O(n) |
| Optimized ⭐ | Single pass + variable | O(1) |

---

```python
def productExceptSelf_v1(nums):
    n = len(nums)
    left = [1]*n
    right = [1]*n

    for i in range(1, n):
        left[i] = left[i-1] * nums[i-1]

    for i in range(n-2, -1, -1):
        right[i] = right[i+1] * nums[i+1]

    return [left[i]*right[i] for i in range(n)]
```

```python
def productExceptSelf_v2(nums):
    n = len(nums)
    output = [1]*n

    for i in range(1, n):
        output[i] = output[i-1] * nums[i-1]

    right = 1
    for i in range(n-1, -1, -1):
        output[i] *= right
        right *= nums[i]

    return output
```

```python
# Approach 1: Prefix & Suffix

nums = [1, 2, 3, 4]
result = productExceptSelf_v1(nums)

print("Approach 1")
print("Input :", nums)
print("Output:", result)

# Test
assert result == [24, 12, 8, 6]

print("Test Passed ")
```

```python
# Approach 2: Optimized O(1) Space

nums = [1, 2, 3, 4]
result = productExceptSelf_v2(nums)

print("Approach 2")
print("Input :", nums)
print("Output:", result)

# Test
assert result == [24, 12, 8, 6]

print("Test Passed ")
```

## 🧩 Problem 2 — Set Matrix Zeroes  

🔗 https://leetcode.com/problems/set-matrix-zeroes/

### 💡 Idea
If a cell is 0 → set entire row & column to 0  
Modify matrix in-place.

---

### ⚙️ Approaches

This problem involves modifying a matrix such that if any element is zero, its entire row and column must also be set to zero.

Two approaches are commonly used to solve this problem.

---

### Approach 1: Using Additional Memory

In this method, the matrix is traversed to identify all rows and columns that contain at least one zero.

These row and column indices are stored separately.

In a second pass, the matrix is updated by setting all identified rows and columns to zero.

This approach is simple to implement but requires additional memory proportional to the size of the matrix.

---

### Approach 2: In-Place Marker Technique

This approach eliminates the need for extra memory by using the first row and first column of the matrix as markers.

During the initial traversal, whenever a zero is found, the corresponding row and column are marked.

A second pass is then used to update the matrix based on these markers.

Finally, the first row and first column are handled separately to ensure correctness.

This method is more space-efficient and is considered the optimal solution.

---

| Approach | Idea | Space |
|---------|------|------|
| Using Sets | Track rows & cols | O(m+n) |
| Markers ⭐ | Use first row/col | O(1) |

---


```python
def setZeroes_v1(matrix):
    rows, cols = set(), set()
    m, n = len(matrix), len(matrix[0])

    for i in range(m):
        for j in range(n):
            if matrix[i][j] == 0:
                rows.add(i)
                cols.add(j)

    for i in range(m):
        for j in range(n):
            if i in rows or j in cols:
                matrix[i][j] = 0
```

```python
def setZeroes_v2(matrix):
    m, n = len(matrix), len(matrix[0])

    first_row = any(matrix[0][j] == 0 for j in range(n))
    first_col = any(matrix[i][0] == 0 for i in range(m))

    for i in range(1, m):
        for j in range(1, n):
            if matrix[i][j] == 0:
                matrix[i][0] = 0
                matrix[0][j] = 0

    for i in range(1, m):
        for j in range(1, n):
            if matrix[i][0] == 0 or matrix[0][j] == 0:
                matrix[i][j] = 0

    if first_row:
        for j in range(n):
            matrix[0][j] = 0

    if first_col:
        for i in range(m):
            matrix[i][0] = 0
```

```python
# Approach 1: Using Sets

matrix = [
    [1,1,1],
    [1,0,1],
    [1,1,1]
]

setZeroes_v1(matrix)

print("Approach 1")
print("Output:")
for row in matrix:
    print(row)

# Test
expected = [
    [1,0,1],
    [0,0,0],
    [1,0,1]
]

assert matrix == expected

print("Test Passed ")
```

```python
# Approach 2: Using Sets

matrix = [
    [1,1,1],
    [1,0,1],
    [1,1,1]
]

setZeroes_v2(matrix)

print("Approach 2")
print("Output:")
for row in matrix:
    print(row)

# Test
expected = [
    [1,0,1],
    [0,0,0],
    [1,0,1]
]

assert matrix == expected

print("Test Passed ")
```
