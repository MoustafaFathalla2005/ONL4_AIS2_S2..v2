## Problem 1 — Product of Array Except Self

🔗 https://leetcode.com/problems/product-of-array-except-self/

**Task:**  
Return an array where each element is the product of all other elements.

**Constraints:**
- No division
- O(n) time
- Follow-up: O(1) space

### Approach 1 — Prefix & Suffix Arrays

Idea:
- Build left and right products
- Multiply them

Simple but uses extra space

```python
def productExceptSelf_v1(nums):
    n = len(nums)
    left = [1]*n
    right = [1]*n

    for i in range(1, n):
        left[i] = left[i-1] * nums[i-1]

    for i in range(n-2, -1, -1):
        right[i] = right[i+1] * nums[i+1]

    return [left[i]*right[i] for i in range(n)]
```

### Approach 2 — Optimized (O(1) Space)

Idea:
- Store prefix in output
- Use variable for suffix

Best solution for interviews

```python
def productExceptSelf_v2(nums):
    n = len(nums)
    output = [1]*n

    for i in range(1, n):
        output[i] = output[i-1] * nums[i-1]

    right = 1
    for i in range(n-1, -1, -1):
        output[i] *= right
        right *= nums[i]

    return output
```

```python
# Approach 1: Prefix & Suffix

nums = [1, 2, 3, 4]
result = productExceptSelf_v1(nums)

print("Approach 1")
print("Input :", nums)
print("Output:", result)

# Test
assert result == [24, 12, 8, 6]

print("Test Passed ")
```

```python
# Approach 2: Optimized O(1) Space

nums = [-1, 1, 0, -3, 3]
result = productExceptSelf_v2(nums)

print("Approach 2")
print("Input :", nums)
print("Output:", result)

# Test
assert result == [0, 0, 9, 0, 0]

print("Test Passed ")
```

## Problem 2 — Set Matrix Zeroes

🔗 https://leetcode.com/problems/set-matrix-zeroes/

**Task:**  
If a cell = 0 → set its row & column to 0

**Goal:**  
Do it in-place

### Approach 1 — Using Sets

Idea:
- Track rows and columns with zeros
- Update matrix

Easy but not optimal

```python
def setZeroes_v1(matrix):
    rows, cols = set(), set()
    m, n = len(matrix), len(matrix[0])

    for i in range(m):
        for j in range(n):
            if matrix[i][j] == 0:
                rows.add(i)
                cols.add(j)

    for i in range(m):
        for j in range(n):
            if i in rows or j in cols:
                matrix[i][j] = 0
```

### Approach 2 — Optimal (O(1) Space)

Idea:
- Use first row & column as markers

Most efficient solution

```python
def setZeroes_v2(matrix):
    m, n = len(matrix), len(matrix[0])

    first_row = any(matrix[0][j] == 0 for j in range(n))
    first_col = any(matrix[i][0] == 0 for i in range(m))

    for i in range(1, m):
        for j in range(1, n):
            if matrix[i][j] == 0:
                matrix[i][0] = 0
                matrix[0][j] = 0

    for i in range(1, m):
        for j in range(1, n):
            if matrix[i][0] == 0 or matrix[0][j] == 0:
                matrix[i][j] = 0

    if first_row:
        for j in range(n):
            matrix[0][j] = 0

    if first_col:
        for i in range(m):
            matrix[i][0] = 0
```

```python
# Approach 1: Using Sets

matrix = [
    [1,1,1],
    [1,0,1],
    [1,1,1]
]

setZeroes_v1(matrix)

print("Approach 1")
print("Output:")
for row in matrix:
    print(row)

# Test
expected = [
    [1,0,1],
    [0,0,0],
    [1,0,1]
]

assert matrix == expected

print("Test Passed ")
```

```python
# Approach 2: Using First Row & Column as Markers (O(1) Space)

matrix = [
    [0,1,2,0],
    [3,4,5,2],
    [1,3,1,5]
]

setZeroes_v2(matrix)

print("Approach 2")
print("Output:")
for row in matrix:
    print(row)

# Test
expected = [
    [0,0,0,0],
    [0,4,5,0],
    [0,3,1,0]
]

assert matrix == expected

print("Test Passed ")
```
